## Dot product between matrices and vectors

A matrix is the bigger element in linear algebra that often represents the complex or collected data. Mathematically, a matrix is a transformation of unit vectors. Unit vectors, unlike the typical definition on high school textbooks such as a vector that has magnitude of 1, form each column of matrices here. When you apply the collection of unit vectors (a.k.a. matrices) to another vector, you will cause another linear transformation, which is called **dot product**.

```python
# The important part of `np.array()` is... it will transform your unit vectors in a wrong way.
# It will treat your column vectors like row vectors. Think about how we define them.

unit1 = np.array([2, 3]) # When we write a code, it is a row, not a column...
```

```python
# Thus, if you want to construct a matrix from each unit vector, please 'transpose' it.
# 'Transpose' means you just make a row vector into a column vector, and vice versa.
unit2 = np.array([2, -1])
matrix = np.array([unit1, unit2]).transpose()
print(matrix) # It will print out the intended matrix that you wrote down on paper.
```

```python
# Actually, this works exactly the same if you handle the data
# like the one that you wrote down on your paper directly...
simple = np.array([[2, 2],
                  [3, -1]])
print(simple) # But there will be many different circumstances, so I would say, know both!
```

```python
# np.ndim() specifies the number of dimensions.
print(np.ndim(simple))
print(np.ndim(mysecondvector)) # Why 2D vector is showing the dimension 1?
# [1, 2, 3] -> This becomes 1D in NumPy and Data Science.
# Just the idiom difference between Mathematics and Computer science... Bear with me.
```

```python
# np.shape() specifies the number of elements along each dimension.
print(np.shape(simple))
print(np.shape(mysecondvector))
```

```python
another_matrix = np.array([[1, 2, 3],
                           [4, 5, 6]])
print(np.ndim(another_matrix))
print(np.shape(another_matrix)) # (how many rows?, how many columns?)
```

```python
# To perform dot product, use this.
vec = np.array([2, 1])

result = matrix.dot(vec)
print(result) # See how the one on paper is printed out.
```

```python
# In the industry, rather than dot(), we use '@' or 'np.matmul()' the most.
same_result = matrix @ vec
another_same_result = np.matmul(matrix, vec)

print(same_result)
print(another_same_result)
```

As you can see, we are playing around with the unit vectors and the vectors that has the same dimension. All of them are mathematically 2D vectors. What if we combine the elements with different dimensions?

The result of it will **expand that vector space into fewer or more dimensions**. It means that when you **perform operations with non-square matrices** rather than square matrices, you can see there is more dynamics in your number system.

```python
cool_matrix = np.array([[1, 2, 3, 4],
                        [5, 6, 7, 8]])
cool_vector = np.array([1, 2, 3, 4])
print(cool_matrix @ cool_vector) # Dimension is got smaller, can you see?
```

```python
cool_matrix = np.array([[1, 2, 3], 
                        [5, 6, 7]])
cool_vector = np.array([1, 2, 3, 4])
print(cool_matrix @ cool_vector)

# There is an error here. Why? To perform the dot product, the number of columns in the matrix
# MUST match with the number of elements in the vector.
```

## Matrix multiplication

Speaking of `np.matmul()`... What is 'matmul'? This is more advanced transformation happens in a vector space.

The specific dot product that we apply between two matrices are called matrix multiplication. How does it look on coding environment?

```python
# When you get used to Python, you can write the code more elegantly(?)...
print(np.array([[1, 2, 3],
                [4, 5, 6]])
      @ np.array([[1, 1], [2, 2], [3, 3]]))
# In this case, the number of rows and the number of columns must be matched.
# Also, the result will be the matrix of (the number of rows) x (the number of columns) from the previous two matrices, respectively.
```

## Matrices and their properties

There are useful matrices in mathematics so mathematicians decided to call them by special names, such as an identity matrix, an inverse matrix, etc.

On the coding side, `NumPy` kindly provides the essential operations (like `@`, `np.matmul()`, `np.dot()`) so that we don't need to calculate everything by hand. Imagine you are dealing with 5 x 11 matrices with 4 x 5 matrices to calculate matrix multiplication. You will go mad.

Like we saw, **determinant** is the special property of a matrix, demonstrating how much a covered area in a vector space by certain group of unit vectors in scale after linear transformation. It is crucial for many Machine learning algorithms. It is just hidden behind the library you will use such as `sklearn` and beyond. However, that doesn't mean that you can just rely on them forever. After we know the basics, we wouldn't be afraid of any technical jargons that ML paper are saying.

Because it is super important and will be used so many time, calculating determinant on `NumPy` is straightforward.

```python
from numpy.linalg import det # NumPy has special package to support Linear algebraic operations.
```

```python
# Get the determinant of A. This is the result of scaling.
baby = np.array([[1, 0],
                 [0, 1]])
print(det(baby))
A = np.array([[2, 0],
              [0, 3]]) # 2 i-hats and 3 j-hats
print(det(A)) # Determinant grew from 1.0 to 6.0.
```

```python
i_hat = np.array([1, 0])
j_hat = np.array([0, 1])
# Where j-hat is sheared...
j_hat_smeared = np.array([1, 1])
basis = np.array([i_hat, j_hat]).transpose()
basis_after_smearing = np.array([i_hat, j_hat_smeared]).transpose()

print(det(basis))
print(det(basis_after_smearing))
# Smearing doesn't change the area that the unit vectors are covering, so determinant is the same.
```

```python
# This example is about the situation when inversion happened,
# that i-hat and j-hat swap in clockwise positions.

print(det(np.array([[-2, 1],
                    [1, 2]]))) # It makes a negative determinant.
```

When you know the determinant, we can finally calculate the **inverse matrix** of the matrix. But on the coding side, of course, `NumPy` has the solution for you.

```python
from numpy.linalg import inv
```

```python
A = np.array([[4, 2, 4],
              [5, 3, 7],
              [9, 3, 6]])
B = np.array([44, 56, 72])

# X = A's inverse * B
X = inv(A) @ B

print(X) # Thus, following the solution, x = 2, y = 34, z = -8
```

```python
print(inv(A))
```

```python
# Calculating n x n matrices' inverse is complicated, but the computer will do the job for you.
tricky = np.array([[3, 0, 2],
                   [2, 0, -2],
                   [0, 1, 1]])
print(det(tricky))
print(inv(tricky)) # You can try calculate this on paper using Gauss-Jordan method.
```

## Eigenvalues and eigenvector

Now we can finally learn eigenvalues and eigenvector, which are crucial members of numerous machine learning tasks, such as linear regression.

This is one of the most trickiest part to understand the pure calculation. But if you practice simultaneous equation and factorisation, it is doable. However, during the real-world machine learning tasks, of course we cannot calculate everything by hand. However, we must know the basic things so that we don't rely on the black box of borrowing libraries.

```python
# On the code library, NumPy will do everything for you.
from numpy.linalg import eig
```

```python
A = np.array([[-6, 3],
              [4, 5]])

eigenvalues, eigenvectors = eig(A)
print(f'eigenvalues: {eigenvalues}')
print(f'eigenvectors: {eigenvectors}')
```
